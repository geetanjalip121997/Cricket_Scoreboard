```python
def strike_rate(runs,balls):
    return runs/balls *100

#print (strike_rate(50,50))


def is_half_century(runs):
    return runs >= 50
    
#print((is_half_century(100)))

def batter_summery(name,runs,balls):
    print(name,"scored",runs,"runs")
    print("Strike Rate =" , strike_rate(runs,balls))
    
    if is_half_century(runs):
        print("Half century")
    else:
        print("Batter luck next time for half century")



(batter_summery("virat",100,50))
(batter_summery("Rohan",20,20))

```
